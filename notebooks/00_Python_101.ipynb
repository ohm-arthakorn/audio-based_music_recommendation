**E3 ฟังก์ชันแปลงโน้ตเป็นความถี่**: เขียน midi_to_hz(n) จากสูตร

$$f = 440 \times 2^{(n-69)/12}$$

✅ ทดสอบ: 69 → 440.0, 81 → 880.0, 60 → 261.63 (โน้ต C กลาง)
ใบ้: ยกกำลังใน Python ใช้ ** ไม่ใช่ ^ ลองดูว่า 2 ^ 3 ให้ผลเท่าไรและทำไม

**ตอบ** : 2^3 คือ การ XOR ในทาง Binary มันจะทำการแปลงเลข 2 เป็น Binary คือ (10) และ 3 เป็น (11)
XOR จะเปรียบเทียบทาง Logics ว่าต่างกันไหม ? จะให้ผลลัพธ์เป็น 1 : จะให้ผลลัพธ์เป็น 0 (ลักษณะการเขียนแบบ Ternary Operator)

```python
def midi_to_hz(n):
    frequency = 440 * (2 ** ((n-69)/12))
    return frequency

print(f"frequency at n = 69 {midi_to_hz(69)} Hz")
print(f"frequency at n = 81 {midi_to_hz(81)} Hz")
print(f"frequency at n = 60 {round(midi_to_hz(60),2)} Hz")
```

### **การเขียนโปรแกรมเชิงวัตถุ (Object Oriented Programming - OOP)**
___
การเขียนโปรแกรมเชิงวัตถุ หรือ Object Oriented Programming - OOP เป็นการเขียนโปรแกรมที่มองการเขียนโปรแกรมเป็นเหมือนวัตถุ (Object) โดยจะเรียกว่า Class ที่มีคุณสมบัติ (Attributes)และพฤติกรรม (Method) ที่แตกต่างกันไป โดยการเขียนโปรแกรมเชิงวัตถุที่ดี Class จะต้องชัดเจน และให้ Class นั้น ๆ เป็นตัวแทนของสิ่ง ๆ เดียว เช่น Class ของ Product ก็จะต้องเก็บสิ่งที่เกี่ยวข้องกับ Product เท่านั้น 



```python
# Example of create class in OOP
class Human:
    # Constructor has a function to create Property for class.
    def __init__(self, name, address):
        # name and address are called "Property" or "Attribute". It's like a variables.
        self.Name = name    
        self.Address = address

    # This call "Method". It's like a normal function but when it in the class, we call method.
    def Speak(text):
        print(text)
```

ในการเรียนเขียนโปรแกรมแบบ OOP จะมี concept ของการเขียนโปรแกรมทั้งหมด 4 แบบประกอบด้วย
1. Encapsulation
2. Inheritance 
3. Abstract 
4. Polymorphism

#### **1. Encapsulation การห่อหุ้ม**
เป็น concept แรก ๆ ของการเขียนโปรแกรมแบบ OOP เป็นการเขียนโปรแกรมโดยกำหนดให้ Property และ Method สามารถเข้าถึงได้ หรือไม่สามารถเข้าถึงได้ 
ระดับของการเข้าถึงก็จะมีทั้งหมด 3 รูปแบบประกอบด้วย\
1. Publics  มีความหมายว่า ไม่ว่าคำสั่งใด ๆ ก็สามารถเข้าถึงได้ 
2. Protected มีความหมายว่า จะต้องเป็น Child Class เท่านั้น ถึงจะสามารถเข้าถึงได้ 
> Child Class คืออะไร เดี๋ยวเราจะได้พบกันในส่วนของ Inheritance
3. Private มีความหมายว่า ไม่สามารถเข้าถึงจากที่ไหนได้เลย มีเพียงตัวมันเองเท่านั้น ที่เข้าถึงได้

```python
# Example of encapsulation program
class DigitalSafe:
    def __init__(self, pin_code: str, secret_item: str):
        self.__pin_code = pin_code  # Private can assign with __ (double underscore) after declare self.
        self.__secret_item = secret_item
        self.__is_locked = True

    def Unlock(self, pin_code): # Using private attributes have to put keyword `self` in arguments of function
        if pin_code == self.__pin_code:
            self.__is_locked = False
        else:
            print("Pin Code is incorrect !")

    def Lock(self):
        if self.__is_locked == True:
            print("This safe has locked already.") 
        else:
            print("This safe was lock.")

    def get_item(self):
        if self.__is_locked == False:
            return self.__secret_item
        else:
            print("Access Denied.")

    def change_pin(self, old_pin, new_pin):
        if (old_pin == self.__pin_code) and (len(new_pin)==4):
            self.__pin_code = new_pin
            print("Password changes successful.")
        else:
            if old_pin != self.__pin_code:
                print("Password was not correct !")
            else:
                print("Password should be only 4 characters !")
```
