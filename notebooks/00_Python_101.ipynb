**E1 จำนวน sample**: ตั้งตัวแปร `sr = 22050` และ `duration = 2.5 (วินาที)` แล้วคำนวณว่าเสียงยาวเท่านี้มีกี่ sample
✅ ต้องได้ 55125 แล้วลองคิดต่อว่าถ้า duration = 0.01 ได้ทศนิยมหรือไม่ และจำนวน sample ควรเป็นทศนิยมได้หรือเปล่า

**ตอบ** : จำนวน sample ไม่ควรเป็นทศนิยม เพราะว่า การ Sampling คือ การเปลี่ยนจากสัญญาณ Analog มาเป็นสัญญาณ Digital ซึ่งเป็นสัญญาณไม่ต่อเนื่องทางเวลา (Discrete-time signal) ทำให้จะไม่มีระหว่างกลาง เช่น 0.5, 0.25

```python
# ทำแบบฝึกหัดข้อ E1

sr = 22050  # Sampling Rate
duration_1 = 2.5
duration_2 = 0.01

sample1  = sr * duration_1
sample2 = sr * duration_2

print(f"จำนวน sample ที่ duration เท่ากับ {duration_1} วินาที = {sample1}")
print(f"จำนวน sample ที่ duration เท่ากับ {duration_2} วินาที = {round(sample2)}")
```

**E2 อ็อกเทฟ**: สร้าง list ความถี่ [110, 220, 440, 880] แล้ววน loop พิมพ์ความถี่แต่ละตัวพร้อมอัตราส่วนกับตัวก่อนหน้า
✅ อัตราส่วนต้องเป็น 2 ทุกคู่ (ตัวแรกไม่มีตัวก่อนหน้า ต้องจัดการกรณีนี้เอง)

```python
octave = [110, 220, 440, 880]
# แสดงตัวแปรทุกตัวของ list octave สิ่งที่ต้องรู้คือจำนวนของสมาชิกใน list
# มีการใช้คำสั่ง len() เพื่อตรวจสอบจำนวนสมาชิกใน list 
# แล้วใช้คำสั่ง for i in (จำนวนที่ต้องการ) ซึ่งจำนวนที่ต้องการเท่ากับจำนวนในสมาชิกของ list
for i in range(len(octave)):
    if(i==0):
        print(f"Current Frequency = {octave[i]}")
    else:
        print(f"Current Frequency = {octave[i]} and Ratio of previous frequency = {octave[i]/octave[i-1]}")
```

**E3 ฟังก์ชันแปลงโน้ตเป็นความถี่**: เขียน midi_to_hz(n) จากสูตร

$$f = 440 \times 2^{(n-69)/12}$$

✅ ทดสอบ: 69 → 440.0, 81 → 880.0, 60 → 261.63 (โน้ต C กลาง)
ใบ้: ยกกำลังใน Python ใช้ ** ไม่ใช่ ^ ลองดูว่า 2 ^ 3 ให้ผลเท่าไรและทำไม

**ตอบ** : 2^3 คือ การ XOR ในทาง Binary มันจะทำการแปลงเลข 2 เป็น Binary คือ (10) และ 3 เป็น (11)
XOR จะเปรียบเทียบทาง Logics ว่าต่างกันไหม ? จะให้ผลลัพธ์เป็น 1 : จะให้ผลลัพธ์เป็น 0 (ลักษณะการเขียนแบบ Ternary Operator)

```python
def midi_to_hz(n):
    frequency = 440 * (2 ** ((n-69)/12))
    return frequency

print(f"frequency at n = 69 {midi_to_hz(69)} Hz")
print(f"frequency at n = 81 {midi_to_hz(81)} Hz")
print(f"frequency at n = 60 {round(midi_to_hz(60),2)} Hz")
```

```python
num_list = [1,2,3]
print(num_list * 2)

import numpy as np
num_list_np = np.array([1,2,3]) * 2
print(num_list_np)
```
